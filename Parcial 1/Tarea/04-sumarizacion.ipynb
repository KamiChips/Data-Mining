```python
import pandas as pd
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

```python
skew_age = df['Age'].skew()
print(skew_age)

#Esto indica que hay sesgo hacia la derecha:
# hay pocos valores dentro de age
#que sean extremadamente altos, mientras que la mayoria
#son valores bajos. por lo que podemos decir que 
#habian mas jovenes y alguno de edad mayor.
```

```python
kurt_fare = df['Fare'].kurt()
print(kurt_fare)
#Nos dice que tan frecuente aparecen valores muy alejados a la media.
#Signica que la distribución es altamente picuda, ya que la columan tiene 
#datos outliers con valores muy altos en comparación a los demás. Lo que nos
#confirma a los outliers que se encontraron anteriormente.
```

```python
data = pd.crosstab(index=df['Pclass'], columns=df['Survived'])
print(data)

#Podemos decir que la clase de cada pasajero se ve relacionada con 
#si sobrevivieron o no. La clase 3, por ejemplo, tiene muchísimos más
#pasajeros que no lograron sobrevivir. Y los de primera clase tuvieron
#una tasa de supervivencia muchísimo mayor.
```
