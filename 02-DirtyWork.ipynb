```python
import random

random_list = [random.random() for _ in range(5_000_000)]
print("Randome list length: ", len(random_list))
```

```python
def mean(values):
    suma = 0
    cantidad = len(values)
    for dato in values:
        suma += dato
        return suma / cantidad

def desviacion(values):
    sd = 0
    mean_val = mean(values)
    for dato in values:
        sd += (dato - mean_val)**2
    varianza = sd / len(values)
    return varianza **0.5
```

```python
print(f'Std: {desviacion(random_list)}')
```

```python
q1 = df['Fare'].quantile(0.25)
#Aqui estamos sacando el cuartil 1
q3= df['Fare'].quantile(0.75)
#Cuartil 3

iqr = q3 - q1

lower_outlier = q1 - 1.5*iqr
upper_outlier = q3 + 1.5*iqr

outliers = df[(df['Fare'] < lower_outlier) | (df['Fare'] > upper_outlier)]

total = len(outliers)

print(total)
print(outliers)
```
