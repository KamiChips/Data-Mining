```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)
```

```python
df.isnull().sum()

#PREGUNTA A
#Porque Android Ver solo contiene 3 valores nulos, mientras que Rating
#tiene alrededor de 1,500 valores nulos y si los eliminamos, estariamos
#borrando una gran parte del dataset.
```

```python
filas_not_dropped = len(df)
df = df.dropna(subset=['Android Ver', 'Type'])
filas_dropped = len(df)
print(f"Filas antes de: {filas_not_dropped}, Filas despues: {filas_dropped}")
```

```python
df['Rating'] = df['Rating'].fillna(
    df.groupby('Category')['Rating'].transform('median')
)
df.isnull().sum()
```

```python
df['Size']
```

```python
df['Size'] = df['Size'].replace("Varies with device", np.nan)
df['Size']
```
