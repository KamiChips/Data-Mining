```python
import pandas as pd
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

```python
print(df['Survived'].value_counts(normalize=True)* 100) 
#PREGUNTA 1
#Dentro de la tasa de supervivencia solo el 38% de los que se encontraban
#en el titanic sobrevivieron y el 61% no lo logro.
```

```python
df.groupby('Sex')['Survived'].mean()* 100
#PREGUNTA 2
#El 74% de los sobrevivientes fueron mujeres, debido al codigo maritimo
#y el otro 18% fueron hombres
```

```python
q1 = df['Fare'].quantile(0.25)
q3 = df['Fare'].quantile(0.75)

iqr = q3 - q1

upper_outlier = q3 + 1.5 * iqr
outlier = df[df['Fare'] > upper_outlier]

print(len(outlier))
print(outlier['Pclass'].value_counts())

#PREGUNTA 3
#Fueron 116 pasajeros quienes pagaron arriba de la tarifa limite, 
#ademas 104 de ellos eran de la primera clase, 7 de tercera clase y solo 
#5 pertenecian a la segunda clase.
```

```python
#PREGUNTA 4
mean_fare = df['Fare'].mean()
median_fare = df['Fare'].median()
print(median_fare)
print(mean_fare)

#Significa que la mayoria de los pasajeros pago menos de 14 
#y algunos de los pasajeros pagaron muchisimo mas de eso
#y hace que la media sea mayor, porque esos valores influyen mucho
#El modelo puede llegar a considerar mucho las diferencias que tiene
#fare debido a su gran variacion, para poder determinar vecinos,
#lo que reduce la influencia de otras variables.
```

```python
#PREGUNTA 5
n = 150

proporcion = df['Survived'].value_counts(normalize=True)
not_survived = round(n * proporcion[0])
survived = n - not_survived

sample_survived = df[df['Survived'] == 1].sample(n=survived, random_state=42)
sample_not = df[df['Survived'] == 0].sample(n=not_survived, random_state=42)
sample = pd.concat([sample_survived, sample_not])
print(len(sample))
print(sample['Survived'].value_counts(normalize=True))

#Estamos evitando que haya un sesgo de muestreo, si utilizamos el aleatorio
#se pueden llegar a seleccionar demasiados sobrevivientes o no sobrevivientes
#y el modelo podria llegar a aprender algo que no representa la distribución
#original
```
